```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab.patches import cv2_imshow
import skimage.io
```

```python
img_bgr = cv2.imread('KTP.png')
print("--- Jawaban 1 ---")
print("Nilai piksel koordinat (100, 100) BGR:", img_bgr[100, 100])
cv2_imshow(img_bgr)
plt.imshow(img_bgr)
plt.show()
```

- Mengapa Berbeda: OpenCV (cv2.imread) membaca citra dalam format warna BGR, sedangkan Matplotlib (plt.imshow) menggunakan format RGB

- Pembuktian: Nilai piksel pada koordinat [100, 100] tercetak sebagai urutan BGR [B, G, R].

```python
canvas_int16 = np.zeros(shape=(530, 530, 3), dtype=np.int16)
canvas_int32 = np.zeros(shape=(530, 530, 3), dtype=np.int32)

print("\n--- Jawaban 2 ---")
print(f"int16 -> dtype: {canvas_int16.dtype}, min: {canvas_int16.min()}, max: {canvas_int16.max()}, nbytes: {canvas_int16.nbytes} bytes")
print(f"int32 -> dtype: {canvas_int32.dtype}, min: {canvas_int32.min()}, max: {canvas_int32.max()}, nbytes: {canvas_int32.nbytes} bytes")
```

- Perbandingan Properti:
int16: dtype=int16, min=0, max=0, nbytes=1685400 bytes (1.68 MB).
int32: dtype=int32, min=0, max=0, nbytes=3370800 bytes (3.37 MB).

- Tampilan: Tampilannya sama saja (keduanya berupa kanvas hitam polos).

- Ukuran Memori: Berbeda. Memori int32 dua kali lebih besar dibanding int16 karena tipe data int32 mengalokasikan 32 bit (4 byte) per elemen piksel, sedangkan int16 hanya mengalokasikan 16 bit (2 byte).


3. - from google.colab.patches import cv2_imshow berfungsi untuk mengimpor fungsi tampilan pengganti khusus Google Colab.

    - cv2.imshow() bawaan OpenCV tidak dapat dipakai di Colab karena fungsi tersebut mencoba membuka jendela GUI native sistem operasi yang menyebabkan crash pada lingkungan cloud notebook Colab.

```python
img_skimage = skimage.io.imread('KTP.png')
img_converted = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)

print("\n--- Jawaban 4 ---")
print("Nilai piksel (100,100) skimage:", img_skimage[100, 100])
print("Nilai piksel (100,100) cvtColor RGB:", img_converted[100, 100])
```

- Warna Sebenarnya: skimage.io.imread() membaca citra langsung dalam format RGB, begitu pula dengan cv2.cvtColor(img, cv2.COLOR_BGR2RGB). Keduanya menampilkan warna asli/sebenarnya.

- Pembuktian: Nilai piksel keduanya pada koordinat yang sama identik (misalnya [R, G, B]), menandakan tidak ada kanal yang tertukar di antara kedua metode tersebut.

```python
print("\n--- Jawaban 5 ---")
print("Ukuran piksel asli (img.shape):", img_bgr.shape)
fig = plt.figure(figsize=(10, 10))
plt.imshow(img_converted)
plt.show()
print("Ukuran piksel setelah figsize (img.shape):", img_bgr.shape)
```

- Nilai img.shape: Tidak berubah.

- Perbedaan:

  - Mengubah ukuran citra (cv2.resize): Mengubah jumlah piksel/data matrik citra secara permanen, sehingga nilai img.shape berubah.

  - Mengubah ukuran tampilan (figsize): Hanya mengubah skala visual penayangan gambar pada layar (plot), tanpa mengubah matriks dan jumlah piksel asli dari citra tersebut.

```python
import cv2
import matplotlib.pyplot as plt
```

```python
img = cv2.imread('KTP.png')
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
```

```python
img_rb = img_rgb.copy()
img_rb[:, :, 1] = 0
```

```python
img_gb = img_rgb.copy()
img_gb[:, :, 0] = 0
```

```python
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(img_rb)
plt.title('Kombinasi Merah-Biru')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(img_gb)
plt.title('Kombinasi Hijau-Biru')
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt
```

```python
img = cv2.imread('KTP.png')
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
```

```python
flip_vert = cv2.flip(img_rgb, 0)
flip_horiz = cv2.flip(img_rgb, 1)
flip_both = cv2.flip(img_rgb, -1)
```

```python
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(flip_vert)
plt.title('Pencerminan Vertikal')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(flip_horiz)
plt.title('Pencerminan Horizontal')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(flip_both)
plt.title('Pencerminan Keduanya')
plt.axis('off')

plt.tight_layout()
plt.show()
```
